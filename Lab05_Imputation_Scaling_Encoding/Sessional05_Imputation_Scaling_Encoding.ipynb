## FEATURE SCALING

Feature scaling means transforming numerical features so they lie in a similar range.  
It helps many machine learning models learn faster and perform better.

### 1. Min-Max Scaling (Normalization)

**What it is**  
Rescales data to a fixed range, usually 0 to 1.

**Formula**

$
x_{scaled}=\frac{x-x_{min}}{x_{max}-x_{min}}
$

**Use when**

- Features have known minimum and maximum bounds
- Neural networks, KNN, distance-based models
- When you want values strictly inside a small range

**Do NOT use when**

- Dataset contains many outliers
- Min or max may change in future data

### 2. Standard Scaling (Z-score Normalization)

**What it is**  
Transforms data so mean becomes 0 and standard deviation becomes 1.

**Formula**

$
x_{scaled}=\frac{x-\mu}{\sigma}
$

where  
$\mu$ = mean,  
$\sigma$ = standard deviation

**Use when**

- Data roughly follows normal distribution
- Linear regression, logistic regression, SVM, PCA
- Features contain outliers but still need normalization

**Do NOT use when**

- You need bounded values like 0–1
- Data is strongly non-Gaussian and bounded scaling is required

### Quick rule

- Need range 0–1 → use MinMaxScaler
- Need centered data (mean 0, std 1) → use StandardScaler

## ENCODING


### Cell 13: One Hot Encoding


### One-Hot Encoding

### When should we use One-Hot Encoding?

Use it when:

• data contains categorical text values
• categories have NO natural order
• example: gender, city, country, department, color

Do NOT use it for ordered categories like:

low < medium < high

For ordered data, use ordinal encoding instead.

#### What is this code doing?

This code converts categorical (text) columns into numeric columns so that machine learning models can understand them.

Most machine learning algorithms cannot work with text values like
`male`, `female`, `first`, `second`.
They require numbers. One-Hot Encoding solves this problem.

Instead of assigning numbers like 0,1,2 (which may create a false order), it creates new binary columns (0 or 1).

Example

Color column

Red, Blue, Green

becomes

Color_Red Color_Blue Color_Green

1 0 0

0 1 0

0 0 1

### Line-by-line explanation

#### 1. Create encoder object

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
```

`OneHotEncoder()` → tool that performs one-hot encoding.

Parameters:

`sparse_output=False`
Normally the encoder returns a sparse matrix (compressed memory format).
Setting False makes it return a normal NumPy array.
This is easier for students to view and convert into a dataframe.

`handle_unknown="ignore"`
If a new category appears later (for example in test data), the encoder will not crash.
Instead, it safely ignores that unknown value.

#### 2. Apply encoding

```python
encoded = ohe.fit_transform(df[cat_cols])
```

`df[cat_cols]` → selects only categorical columns.

`fit_transform()` does two steps:

• **fit** → learn all category names from the data
• **transform** → convert categories into 0/1 columns

The result is stored in `encoded`.

#### 3. Convert result into dataframe

```python
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(cat_cols))
```

Why needed?

The encoder returns only numbers without column names.
This line creates a proper pandas dataframe.

`get_feature_names_out(cat_cols)` automatically generates readable column names like:

sex_female
sex_male
class_First
class_Second

#### 4. Show first rows

```python
encoded_df.head()
```

Displays the first 5 rows so students can verify the output.


````python
# ohe = OneHotEncoder(sparse_output=False)
# ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown="ignore", dtype=np.int32)
encoded = ohe.fit_transform(df[cat_cols])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(cat_cols))

display(df[cat_cols].head())
display(encoded_df.head())
````

Quick rule to remember

- If sparse gives error → use sparse_output
- If sparse_output gives error → use sparse

### Cell 14: Ordinal Encoding


````python
from sklearn.preprocessing import OrdinalEncoder

ord_enc = OrdinalEncoder()
ord_data = ord_enc.fit_transform(df[['class']])
ord_df = pd.DataFrame(ord_data, columns=['class'])
ord_df.head(10)
````

To see the numeric mapping created for the `class` column after using `OrdinalEncoder`, you need to read the learned categories and build a mapping table.

Use this clean classroom-style code.

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

# fit encoder
ord_enc = OrdinalEncoder()
ord_data = ord_enc.fit_transform(df[['class']])

# encoded dataframe
ord_df = pd.DataFrame(ord_data, columns=['class'])

# ---- show mapping ----
classes = ord_enc.categories_[0]           # original labels
mapping = {cat: i for i, cat in enumerate(classes)}

print("Class mapping:")
print(mapping)

ord_df.head(10)
```

What this prints (example)

If your `class` column contains:

```
First, Second, Third
```

The encoder sorts alphabetically and usually produces:

```
{'First': 0, 'Second': 1, 'Third': 2}
```

If you want a FIXED logical order (recommended for teaching Titanic)

Passenger class has natural ranking:

First > Second > Third

So define it manually:

```python
ord_enc = OrdinalEncoder(categories=[['Third','Second','First']])
ord_data = ord_enc.fit_transform(df[['class']])
```

Now mapping becomes:

```
Third  -> 0
Second -> 1
First  -> 2
```

This preserves the socioeconomic ordering.

Very short teaching tip for students

Automatic encoder → alphabetical order
Manual categories → logical real-world order


### Cell 14: Label Encoder

**What is Label Encoder?**
LabelEncoder converts categorical text labels into numbers. Each unique category gets a distinct integer.

**Example from code above:**

- 'Fail' → 0
- 'Pass' → 1

**When to use:**

- For **target variables** (y) that are categorical.
- For **ordered categorical features** (low, medium, high).

**When NOT to use:**

- On **input features** with no natural order (e.g., names, colors, cities).
- Linear or distance-based models (may interpret numbers as ranked).

Use **One-Hot Encoding** instead for unordered categorical input features.


````python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
# Create a small dataframe
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Result": ["Pass", "Fail", "Pass", "Fail", "Pass"]
}
df_small = pd.DataFrame(data)
# Display the original dataframe
print("Original DataFrame:")
display(df_small)
# Create LabelEncoder object
le = LabelEncoder()
# Encode the target column 'Result'
df_small["Result_encoded"] = le.fit_transform(df_small["Result"])
# Display the encoded dataframe
print("\nAfter Label Encoding:")
display(df_small)
````

## FINAL SUMMARY

- Missing values → fillna() or SimpleImputer
- Scaling → StandardScaler or MinMaxScaler
- Encoding → OneHotEncoder or OrdinalEncoder

| Step                      | Method / Function                | Use Case / Notes                                                                             |
| ------------------------- | -------------------------------- | -------------------------------------------------------------------------------------------- |
| Handling Missing Values   | `fillna()`, `SimpleImputer`      | Fill missing values with mean, median, mode, or a constant                                   |
| Scaling / Normalization   | `StandardScaler`, `MinMaxScaler` | StandardScaler: mean 0, std 1; MinMaxScaler: scale to 0–1                                    |
| Encoding Categorical Data | **OneHotEncoder**                | Nominal categorical variables (no order), e.g., color, city                                  |
|                           | **OrdinalEncoder**               | Ordinal variables (with order), e.g., small/medium/large                                     |
|                           | **LabelEncoder**                 | Target labels in classification or ordinal features                                          |
| When Not to Use           | —                                | OneHot: too many categories; Ordinal: for nominal features; Label: nominal features as input |
