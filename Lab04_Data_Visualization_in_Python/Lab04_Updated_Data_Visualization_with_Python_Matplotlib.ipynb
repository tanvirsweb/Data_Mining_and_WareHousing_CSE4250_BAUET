## 1. Import Libraries


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# sns.set_style('whitegrid')
````

## 2. Basic Line Plot

A line plot shows trends across ordered values.


````python
x = np.arange(1,20,2)
y = (x/10) + 5
print(f"x={x}")
print(f"y={y}")
plt.plot(x,y)
plt.title('Basic Line Plot')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.show()
````

## 3. Styled Line Plot with Labels, Legend, Grid


````python
x = np.linspace(0,10,100)
y = np.sin(x)
# print(f"x={x}")
# print(f"y={y}")
plt.plot(x,y,label='sin(x)',marker='o')
plt.plot(x,y+1,label='sin(x)+1',marker='s')
plt.title('Styled Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.grid(True)
plt.show()
````

In **Matplotlib**, the `marker` parameter controls the **style of points** along your line plot. Here’s a full list of commonly used markers you can try:

| ------ | ------------------ | -------------------------- | --- |

| `'3'`  | Tri left (tripod)  | left marker with 3 prongs  |     |
| `'4'`  | Tri right (tripod) | right marker with 3 prongs |     |

| `'p'`  | Pentagon           | ★                          |     |

| `'_'`  | Horizontal line    | ―                          |     |


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 10)
y = np.sin(x)

markers = ['.', ',', 'o', 'v', '^', '<', '>', '1', '2', '3', '4', 's', 'p', '*', 'h', 'H', '+', 'x', 'D', 'd', '|', '_']

plt.figure(figsize=(10,6))

for i, m in enumerate(markers[:10]):  # Plot first 10 markers for demonstration
    plt.plot(x, y+i, marker=m, label=f'marker="{m}"')  # Offset y for clarity

plt.legend(loc='upper right')
plt.title("Example of Different Markers")
plt.show()
````

### **1. Explanation of your code**

```python
x = np.random.rand(50)
```

- **`np.random.rand(50)`** generates **50 random numbers** uniformly distributed in the range **\[0, 1)**.
- Each number is a **float**.
- Example output:

```python
array([0.5488135 , 0.71518937, 0.60276338, 0.54488318, 0.4236548 , ...])
```

- **Use case:** Random samples, simulations, or initializing arrays for testing.

### **2. Common `numpy.random` functions**

| Function                                               | Description                                           | Example                                                          |
| ------------------------------------------------------ | ----------------------------------------------------- | ---------------------------------------------------------------- |
| `np.random.rand(d0, d1, ...)`                          | Uniform random numbers in \[0,1), shape `(d0,d1,...)` | `np.random.rand(3)` → `[0.548, 0.715, 0.602]`                    |
| `np.random.randn(d0, d1, ...)`                         | Standard normal (mean=0, std=1), shape `(d0,d1,...)`  | `np.random.randn(3)` → `[0.978, -0.151, 0.103]`                  |
| `np.random.randint(low, high=None, size=None)`         | Random integers `[low, high)`                         | `np.random.randint(1, 10, 5)` → `[3, 7, 1, 5, 8]`                |
| `np.random.random(size)`                               | Uniform random numbers in \[0,1), same as `rand`      | `np.random.random(4)` → `[0.23, 0.67, 0.12, 0.89]`               |
| `np.random.choice(a, size=None, replace=True, p=None)` | Randomly choose elements from array `a`               | `np.random.choice([10,20,30], 5)` → `[20,10,10,30,20]`           |
| `np.random.seed(seed)`                                 | Sets the seed for reproducibility                     | `np.random.seed(42)`                                             |
| `np.random.shuffle(x)`                                 | Shuffle array `x` in-place                            | `arr = np.array([1,2,3]); np.random.shuffle(arr)`                |
| `np.random.permutation(x)`                             | Returns a shuffled copy of `x`                        | `np.random.permutation([1,2,3])` → `[3,1,2]`                     |
| `np.random.uniform(low=0.0, high=1.0, size=None)`      | Uniform distribution over `[low, high)`               | `np.random.uniform(5,10,3)` → `[6.2, 7.8, 9.1]`                  |
| `np.random.normal(loc=0.0, scale=1.0, size=None)`      | Normal distribution with mean=loc, std=scale          | `np.random.normal(0, 1, 5)` → `[0.12, -0.34, 1.23, 0.55, -0.87]` |
| `np.random.binomial(n, p, size)`                       | Binomial distribution                                 | `np.random.binomial(10, 0.5, 5)` → `[4, 6, 5, 7, 3]`             |
| `np.random.poisson(lam=1.0, size=None)`                | Poisson distribution                                  | `np.random.poisson(3, 5)` → `[2,3,5,1,4]`                        |

---

### **3. Example Usage**

```python
import numpy as np

np.random.seed(42)  # For reproducibility

# Uniform [0,1)
uniform = np.random.rand(5)
print("Uniform:", uniform)

# Standard normal
normal = np.random.randn(5)
print("Normal:", normal)

# Random integers
integers = np.random.randint(1, 10, 5)
print("Integers:", integers)

# Random choice from array
choice = np.random.choice(['A','B','C'], 5)
print("Choice:", choice)

# Uniform in range [5,10)
uniform_range = np.random.uniform(5, 10, 5)
print("Uniform [5,10):", uniform_range)

# Normal with mean=50, std=10
normal_custom = np.random.normal(50, 10, 5)
print("Normal (50,10):", normal_custom)
```


## 7. Box Plot with Seaborn

Displays median, quartiles, and outliers.


````python
data={
    'Category':np.random.choice(['A','B','C'],100),
    'Values':np.random.randn(100)*10+50
}
df=pd.DataFrame(data)
sns.boxplot(x='Category',y='Values',data=df)
plt.title('Boxplot Example')
plt.grid(False)
plt.show()
````

## 8. Pairplot using Iris Dataset

Shows relationships among all numerical variables.


````python
iris=sns.load_dataset('iris')
sns.pairplot(iris,hue='species')
````

## 9. Heatmap Example


````python
matrix=np.random.rand(5,5)
print(matrix)

sns.heatmap(matrix,annot=True)
plt.title('Heatmap Example')
plt.show()
````

## 10. Subplots (Multiple plots in one figure)


````python
x=np.linspace(0,10,100)
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].plot(x,np.sin(x))
ax[0].set_title('sin(x)')
ax[1].plot(x,np.cos(x))
ax[1].set_title('cos(x)')
plt.show()
````
