### Dropping Duplicates in Pandas

#### Remove all duplicated rows

```python
df = df.drop_duplicates()
df.reset_index(drop=True, inplace=True)
```

**Concept:** After removing rows, the index may become non-sequential (e.g., 0, 2, 5). `reset_index(drop=True, inplace=True)` creates a continuous index starting from 0.

- `drop=True` → discards old index
- `drop=False` → keeps old index as a column

#### Remove duplicates based on specific columns

```python
df = df.drop_duplicates(subset=['col1', 'col2'])
```

#### Control which duplicate instance is retained

```python
# Keep first occurrence
df = df.drop_duplicates(keep='first')

# Keep last occurrence
df = df.drop_duplicates(keep='last')

# Remove all duplicates
df = df.drop_duplicates(keep=False)
```

# Univariate Data Analysis

- Univariate data refers to a type of data in which `each observation` or data point `corresponds to a single variable`. Analyzing univariate data is the simplest form of analysis
  in statistics.. Try to gain knowledge from each
  column

- Suppose that the heights of seven students in a class is recorded, height: [164 167 170 174 178 180 179]. There is `only one variable`, which is height, and it is `not dealing with any cause or relationship`.

- For multiple column dataset each column can be treated as one variable. If we analyze data with two variables then it is
  called bivariate analysis and if include more than two variables then it is called multivariate analysis.

## Types of Data

Generally, two major types of data are available:

### i) Numerical Data

Examples:

- Age
- Height
- Salary

These variables are quantitative and measurable, and they support mathematical and statistical operations such as mean, variance, and correlation.

### ii) Categorical Data

Examples:

- Name
- Department
- City

These variables are qualitative in nature and represent labels, groups, or categories rather than numeric magnitude.

#### Conceptual Understanding of Distribution Plots in Seaborn

##### `distplot()` (deprecated)

This was an earlier Seaborn function that **combined two visualizations** in a single command:

- Histogram → bars representing frequency distribution
- KDE curve → smooth line representing probability density

**Single command = combined distribution plot**

It is now **deprecated** because Seaborn redesigned its visualization architecture for better modularity and flexibility.

**Simple meaning:**
`distplot()` = histogram + smooth density curve in one function

##### KDE (Kernel Density Estimation)

KDE is a statistical method used to estimate the **continuous probability density function** of a variable.
It provides a **smooth distribution curve** instead of discrete bars.

##### Modality of Distribution

Modality describes the **number of peaks (modes)** in a distribution:

- **Unimodal** → one peak
- **Bimodal** → two peaks
- **Multimodal** → more than two peaks

These concepts help identify **data structure, clustering, and population subgroups** in exploratory data analysis.

##### Modern Equivalent

```python
sns.displot(df['age'], kde=True)
```

**Meaning:**
Histogram + KDE curve using the updated Seaborn API.


````python
sns.displot(df['age'], kde=True)
````

````python
sns.histplot(df['age'], kde=True)
````

### c) boxplot()

A boxplot is a graph that uses boxes and lines to show
the distribution of a set of numbers.

- Minimum: The smallest number in the data set
- First quartile: The value at which the first 25% of the data falls
- Median: The middle value in the data set
- Third quartile: The value at which the first 75% of the data falls
- Maximum: The largest number in the data set


````python
sns.boxplot(df['age'])
````

### Skewness

Skewness is a statistical measure that assesses the asymmetry of a probability distribution. It quantifies the extent to which the data is skewed or shifted to one side. Positive skewness indicates a longer tail on the right side of the distribution, while negative skewness indicates a longer tail on the left side.


````python
sns.kdeplot(df['age'])
````

### Skew() function determines the skew value of the distribution.

Zero value means symmeltric distribution, positive value indicates positive skews and negative value indicates negative skews. Here positive value means more aged passengers are less.

Skewness measures the asymmetry of a distribution:

- Skewness = 0 : symmetric distribution

- Skewness > 0 : right-skewed (long tail on the right)

- Skewness < 0 : left-skewed (long tail on the left)


````python
df['age'].skew()
````

````python
print(f"{df['age'].skew():.2f}")
````
